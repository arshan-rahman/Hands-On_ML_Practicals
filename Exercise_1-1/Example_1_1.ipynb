```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

plt.style.use('dark_background')

# 1. Download and prepare the data
data_root = "https://github.com/ageron/data/raw/main/"

lifesat = pd.read_csv(data_root + "lifesat/lifesat.csv")

X = lifesat[["GDP per capita (USD)"]].values
y = lifesat[["Life satisfaction"]].values

# 3. Select a Linear Regression model
model = LinearRegression()

# 4. Train the model
model.fit(X, y)

# 5. Generate predictions for the regression line
X_sorted = np.sort(X, axis=0)
y_pred = model.predict(X_sorted)

# 6. Plot scatter points + best-fitting line
plt.figure(figsize=(10, 6))

plt.scatter(X, y, color="blue", label="Actual Data",s=50
)

plt.plot(
    X_sorted,
    y_pred,
    color="purple",
    linewidth=4,
    label="Fitted Line"
)

plt.xlabel("GDP per capita (USD)")
plt.ylabel("Life satisfaction")
plt.axis([23_500, 62_500, 4, 9])
plt.grid(True)
plt.legend()
plt.show()

# 8. Make a prediction for Cyprus
X_new = [[37_655.2]]

prediction = model.predict(X_new)

print("\nPredicted Life Satisfaction for Cyprus:", prediction[0][0])
```
